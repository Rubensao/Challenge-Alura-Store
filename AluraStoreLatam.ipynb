#1. Análisis de facturación



```python
import pandas as pd

# URLs de las tiendas
urls = [
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv",
    "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"
]

# Cargar dataframes
tiendas = [pd.read_csv(url) for url in urls]

# Calcular facturación por tienda
for i, tienda in enumerate(tiendas, start=1):
    facturacion = tienda['Precio'].astype(float).sum()
    print(f"Facturación Tienda {i}: ${facturacion:.2f}")
```

# 2. Ventas por categoría

```python
# Agrupar por categoría y sumar ventas
for i, tienda in enumerate(tiendas, start=1):
    print(f"\nVentas por categoría - Tienda {i}")
    ventas = tienda.groupby("Categoría del Producto")['Precio'].sum()
    print(ventas.sort_values(ascending=False))
```

# 3. Calificación promedio de la tienda


```python
# Calcular la calificación promedio
for i, tienda in enumerate(tiendas, start=1):
    promedio = tienda['Calificación'].mean()
    print(f"Calificación promedio - Tienda {i}: {promedio:.2f}")
```

# 4. Productos más y menos vendidos

```python
# Calcular ingreso total por producto
for i, tienda in enumerate(tiendas, start=1):
    print(f"\nTienda {i} - Productos más vendidos:")
    top_productos = tienda.groupby("Producto")["Precio"].sum().sort_values(ascending=False)
    print(top_productos.head(3))

    print(f"\nTienda {i} - Productos menos vendidos:")
    print(top_productos.tail(3))
```

# 5. Envío promedio por tienda

```python
# Promedio de costos de envío
for i, tienda in enumerate(tiendas, start=1):
    promedio_envio = tienda["Costo de envío"].mean()
    print(f"Costo promedio de envío - Tienda {i}: ${promedio_envio:.2f}")
```

1. Ingresos Totales por Tienda (barra vertical)

```python
import matplotlib.pyplot as plt

# Ingresos totales
ingresos_totales = [tienda['Precio'].sum() for tienda in tiendas]
nombres_tiendas = [f"Tienda {i+1}" for i in range(len(tiendas))]

# Gráfico 1: Barras de ingresos
plt.figure(figsize=(8, 5))
plt.bar(nombres_tiendas, ingresos_totales)
plt.title("Ingresos Totales por Tienda")
plt.ylabel("Ingresos (USD)")
plt.xlabel("Tiendas")
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

2. Participación porcentual por tienda (gráfico de torta/pie)

```python
# Gráfico 2: Pie de participación
plt.figure(figsize=(6, 6))
plt.pie(ingresos_totales, labels=nombres_tiendas, autopct='%1.1f%%', startangle=90)
plt.title("Participación de Cada Tienda en los Ingresos Totales")
plt.tight_layout()
plt.show()
```

3. Promedio de calificación por tienda (barra horizontal)

```python
# Gráfico 3: Calificación promedio
calificaciones = [tienda["Calificación"].mean() for tienda in tiendas]

plt.figure(figsize=(8, 5))
plt.barh(nombres_tiendas, calificaciones, color='skyblue')
plt.title("Calificación Promedio por Tienda")
plt.xlabel("Promedio de Calificación")
plt.grid(axis='x')
plt.tight_layout()
plt.show()
```
